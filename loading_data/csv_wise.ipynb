```python
# Import required libraries
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Load the dataset from CSV file
# Replace 'path to csv' with the actual path to your data
df = pd.read_csv("path to csv")
```

```python
# ===== STEP 1: PREPARE DATA FOR TRAIN-TEST SPLIT =====

# Separate features (X) and target (y)
# axis=1 means we're dropping columns (not rows)
# Replace 'target_column' with your actual target column name
X = df.drop('target_column', axis=1)  # Features: all columns except target
y = df['target_column']  # Target: only the target column

print(f"Total samples: {len(df)}")
print(f"Number of features: {X.shape[1]}")
print(f"Target distribution:\n{y.value_counts()}\n")

# ===== STEP 2: SPLIT DATA INTO TRAIN AND TEST SETS =====

# Split into 80% train, 20% test
# random_state=42 ensures reproducible results
# stratify=y maintains the same class distribution in both sets (important for classification)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2,  # 20% of data goes to test set
    random_state=42,  # Reproducibility seed
    stratify=y  # Maintain class distribution
)

# ===== DISPLAY SPLIT INFORMATION =====

print("Train-Test Split Results:")
print("-" * 50)
print(f"Training set size: {X_train.shape[0]} samples ({X_train.shape[0]/len(df)*100:.1f}%)")
print(f"Test set size: {X_test.shape[0]} samples ({X_test.shape[0]/len(df)*100:.1f}%)")
print(f"\nTraining set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"\nTraining labels distribution:\n{y_train.value_counts()}")
print(f"\nTest labels distribution:\n{y_test.value_counts()}")
```

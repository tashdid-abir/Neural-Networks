```python
# Import TensorFlow for loading image datasets
import tensorflow as tf
```

```python
# ===== CONFIGURATION PARAMETERS =====
# Image dimensions for resizing all input images
IMAGE_SIZE = (224, 224)

# Number of samples to process in each batch
BATCH_SIZE = 16

# Random seed for reproducibility
SEED = 42
```

```python
# ===== STEP 1: LOAD IMAGE DATASETS FROM FOLDER STRUCTURE =====

# Assumptions:
# Folder structure should be:
#   train_directory/
#     ├── class1/
#     ├── class2/
#     └── class3/
#   test_directory/
#     ├── class1/
#     ├── class2/
#     └── class3/

# Load training dataset
# labels="inferred" means class labels are determined from folder names
# label_mode="int" returns labels as integers (0, 1, 2, ...)
train_ds = tf.keras.utils.image_dataset_from_directory(
    directory=r"path to train directory",
    labels="inferred",
    label_mode="int",
    image_size=IMAGE_SIZE,  # Resize all images to 224x224
    batch_size=BATCH_SIZE,  # Group 16 images per batch
    shuffle=True,  # Randomize order for training
    seed=SEED  # Fixed seed for reproducibility
)

# Load test dataset
# Note: shuffle=False to maintain consistent evaluation
test_ds = tf.keras.utils.image_dataset_from_directory(
    directory=r"path to test directory",
    labels="inferred",
    label_mode="int",
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False,  # No shuffling for evaluation
)

# Load validation dataset
val_ds = tf.keras.utils.image_dataset_from_directory(
    directory=r"path to validation directory",
    labels="inferred",
    label_mode="int",
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False,  # No shuffling for evaluation
)

# Display dataset information
print("\n")
print("Dataset Information:")
print("-" * 50)
print("Training Classes   : ", train_ds.class_names)
print("Test Classes       : ", test_ds.class_names)
print("Validation Classes : ", val_ds.class_names)
```

```python
# ===== STEP 2: EXTRACT TRAINING LABELS =====

# Initialize list to store all training labels
train_labels = []

# Iterate through all batches in the training dataset
# Each batch contains (images, labels)
for _, labels in train_ds:  # _ is used as placeholder for images (not needed)
    # Extract label values from tensor format and append to list
    train_labels.extend(labels.numpy().tolist())

print(f"\nExtracted {len(train_labels)} training labels")
print(f"Unique classes: {sorted(set(train_labels))}")
```
